```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,SimpleRNN
import numpy as np
```

```python
x=np.array(
    [
    [[1], [2], [3]],
    [[2], [3], [4]],
    [[3], [4], [5]],
    [[4], [5], [6]],
    [[5], [6], [7]]
], dtype=np.float32)
```

```python
y=np.array([4,5,6,7,8],dtype=np.float32)
```

```python
model=Sequential([ 
    SimpleRNN(16,activation='tanh',input_shape=(3,1)),
    Dense(1)
]
)
```

```python
model.compile(
    optimizer="adam",
    loss="mse"
)
```

```python
model.fit(x,y,epochs=100,verbose=2)
```

```python
test_input = np.array([[[6], [7], [8]]], dtype=np.float32)
prediction=model.predict(test_input,verbose=1)
```
